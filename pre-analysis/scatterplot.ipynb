```python
# Import pandas for loading and handling the CSV data
# matplotlib configuring plot properties and displaying the canvas
# seaborn for advanced statistical data visualization styles

import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
# read the csv file from relative path 

filepath= "../Video_Games_Sales_as_at_22_Dec_2016.csv"
df = pd.read_csv(filepath)

# removed NaN values for the subset Critic_Score and Global_Sales
df_clean = df.dropna(subset=["Critic_Score", "Global_Sales"]).copy()
```

```python
# Convert the Critic_Score and Global_Score column from text/strings into actual numbers (numeric format)

df_clean["Critic_Score"] = pd.to_numeric(df_clean["Critic_Score"])
df_clean["Global_Sales"] = pd.to_numeric(df_clean["Global_Sales"])
```

```python
# Create a regression plot that displays both the scatter points and a linear trend line
sns.regplot(
    data=df,                            
    x='Critic_Score',                   
    y='Global_Sales',                  
    scatter_kws={'alpha': 0.5, 'color': "#e989aa"}, 
    line_kws={'color': 'black', 'linewidth': 1}     
)

# This section defines the titles, labels, axis limits, and renders the final plot
plt.title('Data Representation: Video Game Global Sales vs. Metacritic Critic Score', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Critic_Score (Metacritic Aggregate)', fontsize=11)
plt.ylabel('Global_Sales (Millions of Units)', fontsize=11)
plt.xlim(0, 105)
plt.ylim(-5, 90)
plt.show()
```
